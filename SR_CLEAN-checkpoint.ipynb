```python
# Churn By plan type 
churn_by_plan = (df.groupby('plan_type')['churn_flag'].mean().mul(100).round(2).reset_index(name= 'churn_rate_pct'))
print(churn_by_plan)
# Churn by state 
```

```python
# Churn by state + sum (revanue) and count of user ---

churn_by_state = (
    df.groupby('state')
    .agg(
        total_users=('customerid', 'nunique'),
        churned_users=('churn_flag', 'sum'),
        churn_rate_pct=('churn_flag', lambda x: round(x.mean() * 100, 2)),
        total_revenue=('monthly_charges', 'sum')
    )
    .reset_index()
    .sort_values('churn_rate_pct', ascending=False)
)

print(churn_by_state)
```

```python
print(round(df['churn_flag'].mean() * 100, 2))
```

```python
#churn by subcription type + sum (revenue) and count o users
churn_by_subscription = (
    df.groupby('subscription_type')
    .agg(
        total_users=('customerid', 'nunique'),
        churned_users=('churn_flag', 'sum'),
        churn_rate_pct=('churn_flag', lambda x: round(x.mean() * 100, 2)),
        total_revenue=('monthly_charges', 'sum')
    )
    .reset_index()
    .sort_values('churn_rate_pct', ascending=False)
)

print(churn_by_subscription)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('customer_churn.db')
raw = pd.read_sql("SELECT * FROM db_support", conn)
conn.close()

raw['complaint_date'] = pd.to_datetime(raw['complaint_date'], errors='coerce')
latest = raw.sort_values('complaint_date').drop_duplicates('customerid', keep='last')

# df mein escalations wapas text (Y/N) ke saath
df['escalations'] = df['customerid'].map(latest.set_index('customerid')['escalations'])

print(df['escalations'].unique())   # [nan, 'Y', 'N'] aana chahiye
```

```python
escalation_rate = (df['escalations'] == 'Y').mean() * 100
print("Escalation Rate =", round(escalation_rate, 2), "%")
print((df['escalations'] == 'Y').sum())   # reference mein 4
```

```python
df['escalation_flag'] = (df['escalations'] == 'Y').astype(int)
print(round(df['escalation_flag'].corr(df['churn_flag']), 2))   # 21 rows par
```

```python
# 11. Churn risk - create a column using existing col
conditions = [
    (df['churn_score'] < 50),
    (df['churn_score'] >= 50) & (df['churn_score'] < 70),
    (df['churn_score'] >= 70)
]

choices = ['low', 'med', 'high']

df['churn_risk'] = np.select(conditions, choices, default='unknown')
```
